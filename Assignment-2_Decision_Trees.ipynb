if-else for the decision tree(Assuming we are adding the employee and attributes to the dataframe and then classifying the team)

```python
def team_classify(row):
    if row["Gender"] == "F":
        return "A"
    
    else:
        if row["Department"]=="Finance":
            return "B"
        
        elif row["Department"]=="Software":
            if row["Income"]>=40000:
                return "A"
            
            else:
                return "B"
        
        else:
            if row["Income"]>=40000:
                return "A"
            
            else:
                return "B"
```

Taking input the attributes of the person and giving the team as output

```python
dept=input("Enter department: ")
gend=input("Enter gender: ")
inc=int(input("Enter income: "))

employee={
    "Department": dept,
    "Gender": gend,
    "Income": inc,
    "Team": np.nan
}

team=team_classify(employee)
employee["Team"]=team
print(f"Team is {team}")

df=pd.concat([df,pd.DataFrame([employee])])
```
